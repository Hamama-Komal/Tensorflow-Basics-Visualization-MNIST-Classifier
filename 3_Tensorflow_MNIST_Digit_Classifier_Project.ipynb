**MNIST Digit Classification with a Simple Neural Network.**

```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
```

```python
mnist = keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
plt.imshow(X_train[0], cmap ='gray')
plt.title(f'label: {y_train[0]}')
plt.show()
```

```python
X_train, X_test = X_train/255, X_test/255
```

```python
# defing model
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),  # Input Layer (FLatten the image)
    keras.layers.Dense(128, activation='relu'),  # Hidden Layer (128 neurons)
    keras.layers.Dense(10, activation='softmax')  # Output Layer (10 classes)
])
```

```python
# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics= ['accuracy'])
```

```python
# train model
model.fit(X_train, y_train, epochs = 5)
```

```python
# test the model
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f'Test accuracy: {test_acc}')
```

```python
# prediction
value = 5
prediction = model.predict(X_test[value:value+1])
print(f'Prediction: {np.argmax(prediction[0])}')

plt.imshow(X_test[value], cmap='gray')
plt.title(f'Predicted: {np.argmax(prediction[0])}')
plt.show()
```
